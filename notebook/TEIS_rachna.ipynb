```python
##  DQ: standardized score used for full domains.

dq = studentevals_df[['Adaptive Developmental Quotient',
                'Social-Emotional Developmental Quotient',
                'Communication Developmental Quotient',
                'Motor Developmental Quotient',
                'Cognitive Developmental Quotient']]
dq_average = dq.mean()
dq_average.sort_values(ascending = False)
```

```python
comparison[['Eligibility Mean', 'Annual Mean']].plot(
    kind='bar',
    figsize=(12, 6),
    color=['steelblue', 'darkorange']
)

plt.title('Mean CSS: Eligibility vs. Annual Evaluations')
plt.ylabel('Mean CSS')
plt.xlabel('CSS Subdomains')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Evaluation Type')
plt.tight_layout()
plt.show()
```

```python
top1_contract1 = (
    contract1_EE.groupby(examiner_col)
    .size()
    .sort_values(ascending=False)
    .head(1)
    .reset_index(name='Total Evaluations')
)


top1_contract2 = (
    contract2_EE.groupby(examiner_col)
    .size()
    .sort_values(ascending=False)
    .head(1)
    .reset_index(name='Total Evaluations')
)

top1_contract3 = (
    contract3_EE.groupby(examiner_col)
    .size()
    .sort_values(ascending=False)
    .head(1)
    .reset_index(name='Total Evaluations')
)

display(top1_contract1)
display(top1_contract2)
display(top1_contract3)
```

```python
## top1 contract2_EE examiner average dq

top1_contract2_dq = (
    contract2_EE[
        contract2_EE[examiner_col].isin(top1_contract2[examiner_col])
    ]
    .groupby(examiner_col)[dq_columns]
    .mean()
    
)
top1_contract2_dq 

```
