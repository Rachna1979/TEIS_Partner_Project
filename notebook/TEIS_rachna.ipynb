## 1. In which domains (and sub-domains) are children performing highest and lowest?

## Domains calculation

```python
##  DQ: standardized score used for full domains.

dq = studentevals_df[['Adaptive Developmental Quotient',
                'Social-Emotional Developmental Quotient',
                'Communication Developmental Quotient',
                'Motor Developmental Quotient',
                'Cognitive Developmental Quotient']]
dq_average = dq.mean()
dq_average.sort_values(ascending = False)
```

## sub-domains calculation

```python
ss = studentevals_df[['Adaptive-Self Care SS', 'Adaptive-Personal Responsibility SS',
         'Social Emotional-Adult Interaction SS', 'Social Emotional-Peer Interaction SS', 'Social Emotional-Self Concept / Social Role SS',
         'Communication-Receptive Communication SS', 'Communication-Expressive Communication SS',
        'Motor-Gross Motor SS', 'Motor-Fine Motor SS', 'Motor-Perceptual Motor SS',
        'Cognitive-Attention and Memory SS', 'Cognitive-Reasoning / Academic Skills SS', 'Cognitive-Perception and Concepts SS']]
ss_average = ss.mean()
ss_average.sort_values(ascending = False)
```

## 2. What is the pattern of children's scores? Do individual children tend to perform similarly across all domains or are there differences from domain to domain? Are these patterns similar for all children?

```python
# Calculate highest DQ score for each evaluation
highest = dq.max(axis=1)

# Calculate lowest DQ score for each evaluation
lowest = dq.min(axis=1)

# Calculate the difference
difference = highest - lowest

# Create a summary
score_patterns = pd.DataFrame({
    'Highest DQ': highest,
    'Lowest DQ': lowest,
    'Difference': difference
})

score_patterns.head(10)
```

```python
## pattern of scores using dq difference

score_patterns['Difference'].describe()
```
